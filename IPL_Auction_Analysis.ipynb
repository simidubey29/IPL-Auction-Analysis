```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/IPL_Auction.csv", sep="\t")

top_players = df.sort_values(
    by="Price",
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top_players["Player"],
    top_players["Price"]
)

plt.xticks(rotation=45)
plt.title("Top 10 Most Expensive IPL Players")
plt.xlabel("Player")
plt.ylabel("Price (Crores)")
plt.tight_layout()
plt.show()
```

```python
team_spending = (
    df.groupby("Team")["Price"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

team_spending.plot(kind="bar")

plt.title("Total Team Spending")
plt.xlabel("Team")
plt.ylabel("Total Spending (Crores)")
plt.tight_layout()
plt.show()
```

```python

role_analysis = (
    df.groupby("Role")["Price"]
    .mean()
)

plt.figure(figsize=(7,7))

role_analysis.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Average Spending by Role")
plt.ylabel("")
plt.show()
```

```python
player_count = df["Team"].value_counts()

plt.figure(figsize=(10,6))

player_count.plot(kind="bar")

plt.title("Number of Players Per Team")
plt.xlabel("Team")
plt.ylabel("Player Count")
plt.tight_layout()
plt.show()
```

```python
highest_paid = df.loc[
    df.groupby("Team")["Price"].idxmax()
]

plt.figure(figsize=(12,6))

plt.bar(
    highest_paid["Player"],
    highest_paid["Price"]
)

plt.xticks(rotation=45)
plt.title("Highest Paid Player of Each Team")
plt.xlabel("Player")
plt.ylabel("Price (Crores)")
plt.tight_layout()
plt.show()
```

```python
batsmen = df[df["Role"] == "BAT"]
top_batsmen = batsmen.sort_values(
    by="Price",
    ascending=False
).head(5)

plt.figure(figsize=(10,5))

plt.bar(
    top_batsmen["Player"],
    top_batsmen["Price"]
)

plt.title("Top 5 Most Expensive Batsmen")
plt.xlabel("Player")
plt.ylabel("Price (Crores)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bowlers = df[df["Role"] == "BOWL"]

top_bowlers = bowlers.sort_values(
    by="Price",
    ascending=False
).head(5)

plt.figure(figsize=(10,5))

plt.bar(
    top_bowlers["Player"],
    top_bowlers["Price"]
)

plt.title("Top 5 Most Expensive Bowlers")
plt.xlabel("Player")
plt.ylabel("Price (Crores)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python


ar = df[df["Role"] == "AR"]

top_ar = ar.sort_values(
    by="Price",
    ascending=False
).head(5)

plt.figure(figsize=(10,5))

plt.bar(
    top_ar["Player"],
    top_ar["Price"]
)

plt.title("Top 5 Most Expensive All-Rounders")
plt.xlabel("Player")
plt.ylabel("Price (Crores)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
role_count = df["Role"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    role_count,
    labels=role_count.index,
    autopct="%1.1f%%"
)

plt.title("Role Distribution")
plt.show()
```

```python
avg_price = df.groupby(
    "Team"
)["Price"].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10,6))

avg_price.plot(kind="bar")

plt.title("Average Player Price by Team")
plt.ylabel("Average Price (Crores)")
plt.tight_layout()
plt.show()
```

```python


role_spending = df.groupby(
    "Role"
)["Price"].sum()

plt.figure(figsize=(8,5))

role_spending.plot(kind="bar")

plt.title("Total Spending on Each Role")
plt.ylabel("Total Spending")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["Price"],
    bins=10
)

plt.title("Distribution of Auction Prices")
plt.xlabel("Price (Crores)")
plt.ylabel("Number of Players")
plt.show()
```

```python
import os

os.makedirs("charts", exist_ok=True)

def save_chart(filename):
    plt.savefig(
        f"charts/{filename}",
        dpi=300,
        bbox_inches="tight"
    )
```
